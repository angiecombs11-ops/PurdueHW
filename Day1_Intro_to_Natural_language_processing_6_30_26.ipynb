```python
#!pip install nltk
import nltk
nltk.download('all')
```

```python
t_text = "To further elaborate on the geographical trends, North America has procured more than 50% of the global share in 2017 and has been leading the regional landscape of AI in the retail market.** The U.S. has a significant credit in the regional trends with over 65% of investments (including M&As, private equity, and venture capital) in artificial intelligence technology. Additionally, the region is a huge hub for startups in tandem with the presence of tech titans, such as Google, IBM, and Microsoft"
```

```python
t_text
```

```python
print(f"Length of the text {len(t_text)} characters")
```

Sentence Tokenization

```python
from nltk.tokenize import sent_tokenize
```

```python
sentences = sent_tokenize(t_text)
```

```python
for sent in sentences:
    print(sent)
    #print("\n")
```

Word Tokenization

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(t_text)
words[0:10]
```

```python
len(words)
```

Convert to lower

```python

words = list(map(str.lower, words))
```

```python
words[0:10]
```

```python
len(words)
```

Remove Duplicate

```python
# Remove duplicate words
words = list(set(words))
```

```python
len(words)
```

Remove punctuation and numbers

```python
words_in_para = []
for word in words:
    if word.isalpha():
        words_in_para.append(word)
```

```python
words_in_para[0:10]
```

```python
len(words_in_para)
```

Stopwords

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')[0:10]
```

```python
new_words = []
for word in words_in_para:
    if word not in stopwords.words('english'):
        new_words.append(word)
```

```python
len(new_words)
```

```python
new_words[0:10]
```

Stemming and Lemmatization

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
ps.stem('cycling')
```

```python
ps.stem('cycles')
```

```python
ps.stem('cycled')
```

```python
stemmed_words = []
for word in new_words:
    print(f'Original Word {word} , Stem : {ps.stem(word)}')
    if ps.stem(word) not in stemmed_words:
        stemmed_words.append(ps.stem(word))
```

```python
len(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
wn = WordNetLemmatizer()
```

```python
wn.lemmatize("cycle",'v')
```

```python
stemmed_words[0:10]
```

N-Grams

```python
ng = nltk.ngrams("The quick brown fox jumps over the lazy dog".split(),3)

for a in ng:
    print(a)
```

Parts of Speech Tagging

```python
nltk.pos_tag("John and Mary go to the school".split())
```

```python
nltk.pos_tag(new_words)
```

Named Entity Resolution

```python
print(nltk.ne_chunk(nltk.pos_tag(word_tokenize(t_text)), binary=True))
```

Using Spacy

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(t_text)
```

```python
for token in doc:
    print(f"{token.text:{15}} {token.tag_:{10}} {token.pos_:{10}} {spacy.explain(token.tag_)}")
```

```python
new_words_dict = {}
for token in doc:
    if token.text.lower() not in stopwords.words('english'):
        if token.text.isalpha():
            new_words_dict[token.text.lower()] = token.pos_
            print(f"{token.text:{15}} {token.tag_:{10}} {token.pos_:{10}} {spacy.explain(token.tag_)}")
```
